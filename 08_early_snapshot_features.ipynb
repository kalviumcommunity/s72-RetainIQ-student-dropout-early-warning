```python
import pandas as pd

students = pd.read_csv("data/raw/students.csv")
sessions = pd.read_csv("data/raw/sessions.csv")
quizzes = pd.read_csv("data/raw/quizzes.csv")
completions = pd.read_csv("data/raw/completions.csv")

students["enrollment_date"] = pd.to_datetime(students["enrollment_date"])
sessions["session_date"] = pd.to_datetime(sessions["session_date"])
quizzes["quiz_date"] = pd.to_datetime(quizzes["quiz_date"])
```

```python
CUTOFF_DAYS = 30

sessions_early = sessions.merge(students[["student_id", "enrollment_date"]], on="student_id")
sessions_early["days_since_enroll"] = (sessions_early["session_date"] - sessions_early["enrollment_date"]).dt.days
sessions_early = sessions_early[sessions_early["days_since_enroll"] <= CUTOFF_DAYS]

quizzes_early = quizzes.merge(students[["student_id", "enrollment_date"]], on="student_id")
quizzes_early["days_since_enroll"] = (pd.to_datetime(quizzes_early["quiz_date"]) - quizzes_early["enrollment_date"]).dt.days
quizzes_early = quizzes_early[quizzes_early["days_since_enroll"] <= CUTOFF_DAYS]
```

```python
import pandas as pd
import joblib
import os
from sklearn.metrics import accuracy_score, precision_score, recall_score, roc_auc_score

X_test = pd.read_csv("data/processed/X_test.csv")
y_test = pd.read_csv("data/processed/y_test.csv").squeeze()

lr_model = joblib.load("models/logistic_regression_v1.pkl")
rf_model = joblib.load("models/random_forest_v1.pkl")

lr_pred = lr_model.predict(X_test)
lr_prob = lr_model.predict_proba(X_test)[:, 1]
rf_pred = rf_model.predict(X_test)
rf_prob = rf_model.predict_proba(X_test)[:, 1]

comparison = pd.DataFrame({
    "Model": ["Logistic Regression", "Random Forest"],
    "Accuracy": [accuracy_score(y_test, lr_pred), accuracy_score(y_test, rf_pred)],
    "Precision": [precision_score(y_test, lr_pred), precision_score(y_test, rf_pred)],
    "Recall": [recall_score(y_test, lr_pred), recall_score(y_test, rf_pred)],
    "AUC": [roc_auc_score(y_test, lr_prob), roc_auc_score(y_test, rf_prob)],
})
print(comparison)

os.makedirs("models", exist_ok=True)
final_model = rf_model   # change if Logistic Regression actually wins
joblib.dump(final_model, "models/final_model.pkl")
```
